***Cleaning Trump 2.0 data***

```python
data2025 = pd.read_csv('data/arrests2025clean.csv')
data2025.columns.to_list()
```

```python
columns_to_keep3 = ["apprehension_date", "apprehension_aor", "apprehension_criminality", "unique_identifier", 'birth_year','gender','citizenship_country']
data2025 = data2025[columns_to_keep3]
data2025 = data2025.rename(columns={'apprehension_aor': 'Apprehension AOR', 'unique_identifier': 'Unique Identifier', 'apprehension_date': 'Apprehension Date', 'apprehension_criminality': 'Apprehension Criminality', 'birth_year': 'Birth Year', 'gender': 'Gender', 'citizenship_country': 'Citizenship Country'})
```

```python
#Recoding Criminality Column
data2025 = data2025.replace(to_replace="1 Convicted Criminal", value="Convicted")
data2025 = data2025.replace(to_replace="2 Pending Criminal Charges", value="Pending Charges")
data2025 = data2025.replace(to_replace="3 Other Immigration Violator", value="No Criminal Charges")
```

```python
#Renaming AOR column to be compatible with the topojson
data2025['Apprehension AOR'] = data2025['Apprehension AOR'].str.replace(' Area of Responsibility', '', regex=False)
data2025 = data2025.rename(columns={'Apprehension AOR': 'aor_nam'})
```

```python
data2025['Apprehension MonthYear'] = pd.to_datetime(data2025['Apprehension Date']).dt.strftime('%Y-%m')

#Removing Missing IDs
print("The number of missing IDs in the 2023 to 2025 data is", data2025['Unique Identifier'].isna().sum())
data2025 = data2025.dropna(subset=['Unique Identifier'])

#Checking for Duplicates
data2025[data2025.duplicated(subset=['Unique Identifier', 'Apprehension Date', 'aor_nam'], keep=False)] #6690 duplicates
```

```python
#Removing Duplicates
data2025.drop_duplicates(subset=['Unique Identifier', 'Apprehension Date', 'aor_nam'], inplace=True)
```

```python
#Adding Age - birthdate is redacted
data2025['Apprehension Date'] = pd.to_datetime(data2025['Apprehension Date'])
data2025['Age'] = data2025['Apprehension Date'].dt.year - data2025['Birth Year'].dropna().astype(int)
```

```python
#Not project related - for the portfolio
kmeansdata = data2025.copy()
kmeansdata.to_csv('data/kmeansdata2025.csv', index=False)
```

```python
#Didn't end up using in project, bust still available if you're interested!
arrestsbymonth2025 = data2025.groupby('Apprehension MonthYear').size().reset_index(name='arrests')
arrestsbymonth2025.to_csv('/Users/hannahliu/Desktop/PP434/website/data/arrestsbymonth2025.csv', index=False)
```

***Cleaning Trump 1.0 Data***

```python
data2016 = pd.read_csv('data/arrests2016.csv', header=4)
data2017 = pd.read_csv('data/arrests-2017.csv', header=4)
data2018 = pd.read_csv('data/arrests2018.csv', header=4)
```

```python
columns_to_keep = [
    "Apprehension Date And Time",
    "Apprehension AOR",
    "Numeric Birth Year",
    "Gender",
    "Most Serious Criminal Charge Status",
    "Most Serious Criminal Conviction Charge Code",
    "Citizenship Country",
    "Sequence Number/Unique Identifier"
]

data2017 = data2017[columns_to_keep]
data2016 = data2016[columns_to_keep]
data2018 = data2018[columns_to_keep]

data2017 = pd.concat([data2017, data2016, data2018], ignore_index=True)
full_data_2017 = data2017[columns_to_keep]
```

```python
data2017['Apprehension Date And Time'] = pd.to_datetime(data2017['Apprehension Date And Time']).dt.strftime('%Y-%m-%d')
```

```python
#Cleaning for Uniformity
data2017 = data2017.rename(columns={'Apprehension Date And Time': 'Apprehension Date'})
data2017 = data2017.rename(columns={'Numeric Birth Year': 'Birth Year'})
data2017 = data2017.rename(columns={'Sequence Number/Unique Identifier': 'Unique Identifier'})

#AOR
data2017['Apprehension AOR'] = data2017['Apprehension AOR'].str.replace(' Area of Responsibility', '', regex=False)
data2017 = data2017.rename(columns={'Apprehension AOR': 'aor_nam'})

#Criminality
data2017 = data2017.rename(columns={'Most Serious Criminal Charge Status': 'Apprehension Criminality'})
data2017['Apprehension Criminality'] = data2017['Apprehension Criminality'].fillna('No Criminal Charges')
data2017['Apprehension Criminality'] = data2017['Apprehension Criminality'].replace('Turned Over to INS without Prosecution', 'No Criminal Charges')
data2017['Apprehension Criminality'] = data2017['Apprehension Criminality'].replace('Overturned', 'No Criminal Charges')
data2017['Apprehension Criminality'] = data2017['Apprehension Criminality'].replace('Dismissed', 'No Criminal Charges')
data2017['Apprehension Criminality'] = data2017['Apprehension Criminality'].replace('Pending', 'Pending Charges')
```

**Chart 2: Arrests by Citizenship Country**

```python
arrests_by_country2017 = data2017[
    (data2017['Apprehension Date'] >= '2017-01-20') & 
    (data2017['Apprehension Date'] <= '2017-10-15')
].groupby('Citizenship Country').agg({
    'Unique Identifier': 'count'
}).reset_index()

arrests_by_country2017.columns = ['Citizenship Country', 'Arrests']

#Adding percentage
arrests_by_country2017['Percentage'] = (
    arrests_by_country2017['Arrests'] / arrests_by_country2017['Arrests'].sum() * 100
).round(2)

arrests_by_country2017 = arrests_by_country2017.sort_values('Arrests', ascending=False)
arrests_by_country2017
arrests_by_country2017["Citizenship Country"] = arrests_by_country2017["Citizenship Country"].str.title()
```

***Combining the Datasets for Charts 3-4***

```python
#Select columns in the same order for both datasets
data2025_subset = data2025[['Apprehension Date', 'aor_nam', 'Birth Year', 'Age', 'Gender', 'Apprehension Criminality', 'Citizenship Country', 'Unique Identifier']]
data2025_subset['Administration'] = 'Trump 2.0'
data2017_subset = data2017[['Apprehension Date', 'aor_nam', 'Birth Year', 'Age', 'Gender', 'Apprehension Criminality', 'Citizenship Country', 'Unique Identifier']]
data2017_subset['Administration'] = 'Trump 1.0'

combined_data = pd.concat([data2025_subset, data2017_subset], ignore_index=True)
```

```python
combined_data['Apprehension Date'] = pd.to_datetime(combined_data['Apprehension Date'])
```

***Chart 4: Arrests Views***

```python
combined_data['Age Group'] = pd.cut(combined_data['Age'], 
                                     bins=[0, 18, 25, 35, 45, 55, 65, 100], 
                                     labels=['0-17', '18-24', '25-34', '35-44', '45-54', '55-64', '65+'])
```

```python
full_combined_data = combined_data
```

```python
# Create long format dataset with gender, age group, and criminality groupings
# Filter data to only include Jan 20 to Oct 15 for each administration

# Filter for Trump 1.0 (2017-01-20 to 2017-10-15), excluding Unknown gender
trump1_filtered = combined_data[
    (combined_data['Administration'] == 'Trump 1.0') &
    (combined_data['Apprehension Date'] >= '2017-01-20') &
    (combined_data['Apprehension Date'] <= '2017-10-15') &
    (combined_data['Gender'] != 'Unknown')
]

# Filter for Trump 2.0 (2025-01-20 to 2025-10-15), excluding Unknown gender
trump2_filtered = combined_data[
    (combined_data['Administration'] == 'Trump 2.0') &
    (combined_data['Apprehension Date'] >= '2025-01-20') &
    (combined_data['Apprehension Date'] <= '2025-10-15') &
    (combined_data['Gender'] != 'Unknown')
]

# Initialize empty dataframe
viewdf = pd.DataFrame(columns=['view', 'characteristic', 'value', 'percentage', 'administration'])

# ===== GENDER GROUPING =====
# Trump 1.0 (2017)
gender_2017 = trump1_filtered.groupby('Gender').size().reset_index(name='value')
gender_2017['view'] = 'gender'
gender_2017['characteristic'] = gender_2017['Gender'].str.lower()
gender_2017['administration'] = 'Trump 1.0'
gender_2017['percentage'] = (gender_2017['value'] / gender_2017['value'].sum())
gender_2017 = gender_2017[['view', 'characteristic', 'value', 'percentage', 'administration']]

# Trump 2.0 (2025)
gender_2025 = trump2_filtered.groupby('Gender').size().reset_index(name='value')
gender_2025['view'] = 'gender'
gender_2025['characteristic'] = gender_2025['Gender'].str.lower()
gender_2025['administration'] = 'Trump 2.0'
gender_2025['percentage'] = (gender_2025['value'] / gender_2025['value'].sum())
gender_2025 = gender_2025[['view', 'characteristic', 'value', 'percentage', 'administration']]

viewdf = pd.concat([viewdf, gender_2017, gender_2025], ignore_index=True)
```

```python
# ===== AGE GROUP GROUPING =====
# Trump 1.0 (2017)
agegroup_2017 = trump1_filtered.groupby('Age Group').size().reset_index(name='value')
agegroup_2017['view'] = 'age group'
agegroup_2017['characteristic'] = agegroup_2017['Age Group'].astype(str)
agegroup_2017['administration'] = 'Trump 1.0'
agegroup_2017['percentage'] = (agegroup_2017['value'] / agegroup_2017['value'].sum())
agegroup_2017 = agegroup_2017[['view', 'characteristic', 'value', 'percentage', 'administration']]

# Trump 2.0 (2025)
agegroup_2025 = trump2_filtered.groupby('Age Group').size().reset_index(name='value')
agegroup_2025['view'] = 'age group'
agegroup_2025['characteristic'] = agegroup_2025['Age Group'].astype(str)
agegroup_2025['administration'] = 'Trump 2.0'
agegroup_2025['percentage'] = (agegroup_2025['value'] / agegroup_2025['value'].sum())
agegroup_2025 = agegroup_2025[['view', 'characteristic', 'value', 'percentage', 'administration']]

viewdf = pd.concat([viewdf, agegroup_2017, agegroup_2025], ignore_index=True)
```

```python


# ===== CRIMINALITY GROUPING =====
# Trump 1.0 (2017)
criminality_2017 = trump1_filtered.groupby('Apprehension Criminality').size().reset_index(name='value')
criminality_2017['view'] = 'criminality'
criminality_2017['characteristic'] = criminality_2017['Apprehension Criminality'].astype(str)
criminality_2017['administration'] = 'Trump 1.0'
criminality_2017['percentage'] = (criminality_2017['value'] / criminality_2017['value'].sum())
criminality_2017 = criminality_2017[['view', 'characteristic', 'value', 'percentage', 'administration']]

# Trump 2.0 (2025)
criminality_2025 = trump2_filtered.groupby('Apprehension Criminality').size().reset_index(name='value')
criminality_2025['view'] = 'criminality'
criminality_2025['characteristic'] = criminality_2025['Apprehension Criminality'].astype(str)
criminality_2025['administration'] = 'Trump 2.0'
criminality_2025['percentage'] = (criminality_2025['value'] / criminality_2025['value'].sum())
criminality_2025 = criminality_2025[['view', 'characteristic', 'value', 'percentage', 'administration']]

viewdf = pd.concat([viewdf, criminality_2017, criminality_2025], ignore_index=True)

viewdf
```
