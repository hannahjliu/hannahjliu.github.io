```python
import pandas as pd
import utm
from pyproj import Transformer
```

```python
mapdata = pd.read_csv("/Users/hannahliu/Desktop/PP434/website/data/nrw_authorised_landfill_sites_pnt.csv")

columnstokeep = ["operational_status","site_postcode","site_grid_reference","easting","northing","nrw_area"]
```

```python
mapdata = mapdata[columnstokeep]
```

```python

# Create transformer for British National Grid (EPSG:27700) to WGS84 (EPSG:4326)
transformer = Transformer.from_crs("EPSG:27700", "EPSG:4326", always_xy=True)

def convert_osgb_to_latlong(easting, northing):
    """Convert British National Grid (OSGB36) to Latitude/Longitude (WGS84)"""
    try:
        lon, lat = transformer.transform(easting, northing)
        return pd.Series({'latitude': lat, 'longitude': lon})
    except:
        return pd.Series({'latitude': None, 'longitude': None})

# Apply conversion
mapdata[['latitude', 'longitude']] = mapdata.apply(
    lambda row: convert_osgb_to_latlong(row['easting'], row['northing']), 
    axis=1
)

# Define columns to keep
columnstokeep = [
    'operational_status', 
    'site_postcode', 
    'site_grid_reference', 
    'easting', 
    'northing', 
    'latitude',
    'longitude',
    'nrw_area'
]

# Filter to selected columns
mapdata = mapdata[columnstokeep]

# Save to CSV

# Save to CSV
mapdata.to_csv('/Users/hannahliu/Desktop/PP434/website/data/mapdatawithcoordinates.csv', index=False)
```
