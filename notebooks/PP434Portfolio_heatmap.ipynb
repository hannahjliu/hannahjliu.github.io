```python
import pandas as pd
```

***Cleaning the Data***

```python
data2025 = pd.read_csv('/Users/hannahliu/Desktop/PP434/website/data/arrests2025clean.csv')
columns_to_keep3 = ["apprehension_date", "apprehension_aor", "apprehension_criminality", "unique_identifier", 'birth_year','gender','citizenship_country']
data2025 = data2025[columns_to_keep3]
data2025 = data2025.rename(columns={'apprehension_aor': 'Apprehension AOR', 'unique_identifier': 'Unique Identifier', 'apprehension_date': 'Apprehension Date', 'apprehension_criminality': 'Apprehension Criminality', 'birth_year': 'Birth Year', 'gender': 'Gender', 'citizenship_country': 'Citizenship Country'})
```

```python
data2025['Apprehension AOR'] = data2025['Apprehension AOR'].str.replace(' Area of Responsibility', '', regex=False)
data2025['Apprehension MonthYear'] = pd.to_datetime(data2025['Apprehension Date']).dt.strftime('%Y-%m')
data2025 = data2025.dropna(subset=['Unique Identifier'])
data2025[data2025.duplicated(subset=['Unique Identifier', 'Apprehension Date', 'Apprehension AOR'], keep=False)] #6690 duplicates
data2025 = data2025.drop_duplicates(subset=['Unique Identifier', 'Apprehension Date', 'Apprehension AOR'])
data2025.drop_duplicates(subset=['Unique Identifier', 'Apprehension Date', 'Apprehension AOR'], inplace=True)
data2025 = data2025[data2025['Apprehension Date'] >= '2024-10-01'].copy()

data2025
```

***Generating the Dataset***

```python
data2025['Apprehension Date'] = pd.to_datetime(data2025['Apprehension Date'])
inaug_date = pd.to_datetime('2025-01-20')
data2025['weeks_since_inauguration'] = ((data2025['Apprehension Date'] - inaug_date).dt.days / 7).astype(int)
heatmap = data2025.groupby(['Apprehension AOR', 'weeks_since_inauguration']).size().reset_index(name='arrests')
heatmap

heatmap.to_csv('data/arrests_heatmap_data.csv', index=False)
```
