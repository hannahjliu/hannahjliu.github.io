```python
url = "https://raw.githubusercontent.com/RDeconomist/RDeconomist.github.io/main/charts/library/chartLine1.json"

base_spec = requests.get(url).json()

print(json.dumps(base_spec, indent=4))
```

```python
base_spec['data']
```

```python
base_spec['description'] = 'This is a line chart showing gender equality in government'
```

```python

base_spec["width"] = 200
base_spec["title"]["text"] = "Percent of female members of national parliaments"

## Print out our new Spec:
print(json.dumps(base_spec, indent=2))
```

```python
base_spec['encoding']['x']['field'] = 'date'
base_spec['encoding']['y']['field'] = 'value'
base_spec['data']['url'] = 'XYZ'
base_spec['data']['format']['property'] = '1'
base_spec['encoding']['x']['timeUnit'] = 'year'
base_spec['encoding']['x']['scale'] = {'domain': ['1995', '2025']}
base_spec['encoding']['y']['scale'] = {'domain': [0, 100]}
base_spec['encoding']['y']['title'] = '% of seats held by women'

# Add a reference line at 50%
base_spec['layer'] = [
    {
        'mark': {'type': 'line', 'color': 'tomato', 'strokeWidth': 2, 'opacity': 1},
        'encoding': {
            'x': base_spec['encoding']['x'],
            'y': base_spec['encoding']['y']
        }
    },
    {
        'mark': {'type': 'rule', 'color': 'gray', 'strokeDash': [4, 4]},
        'encoding': {
            'y': {'datum': 50}
        }
    }
]

del base_spec['mark']
del base_spec['encoding']
```

```python
print(json.dumps(base_spec, indent=2))
```

```python
base_api = "https://api.worldbank.org/v2/country/{}/indicator/SG.GEN.PARL.ZS?format=json"
countries = ['swe', 'rwa', 'kor', 'isl', 'nzl', 'tun']

for i in countries:
    apiToUse = base_api.format(i)
  
    spec = copy.deepcopy(base_spec)
    spec['data']['url'] = apiToUse
    spec['title']['subtitle'] = i.upper()

    filename = f'visualizations/{i}_parliament_gender.json'
    with open(filename, 'w') as f:
        json.dump(spec, f, indent=2)
    
    chart = alt.Chart.from_dict(spec)
    display(chart)
```
