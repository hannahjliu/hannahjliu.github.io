```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
```

```python
website_url = "https://secure.ssa.gov/poms.nsf/lnx/0202613900#a1"
req = requests.get(website_url)
```

```python
soup = BeautifulSoup(req.content, "lxml")
print(soup.prettify())
```

```python
tables = soup.find_all("table")
tables = pd.read_html(website_url)

table = tables[3]
print(table)

df = pd.DataFrame(table)
df.to_csv('data/crimeclass.csv', index=False)
```

```python
df = df.rename(columns={'Type of Offense Code  V=violent  D=drug-related  Blank = nonviolent or not drug related': 'Type'})
```
