```python
import pandas as pd
import numpy as np
from datetime import datetime, timedelta

class FamilyMatcher:
    def __init__(self, time_window_hours=4, min_parent_age_gap=16, max_parent_age_gap=50):
        """
        Initialize the family matcher with configurable parameters.
        
        Args:
            time_window_hours: Maximum hours between child and parent book-in times
            min_parent_age_gap: Minimum age difference for parent (default 16)
            max_parent_age_gap: Maximum age difference for parent (default 50)
        """
        self.time_window_hours = time_window_hours
        self.min_parent_age_gap = min_parent_age_gap
        self.max_parent_age_gap = max_parent_age_gap
    
    def load_data(self, filepath_or_df):
        """Load detention data from CSV file or DataFrame."""
        print("Loading data...")
        if isinstance(filepath_or_df, str):
            df = pd.read_csv(filepath_or_df)
        else:
            df = filepath_or_df.copy()
        
        print("Processing columns...")
        # Create standardized columns directly from your specific column names
        df['age'] = datetime.now().year - df['birth_year']
        df['facility'] = df['detention_facility_code_first']
        df['citizenship'] = df['citizenship_country']
        df['stay_id'] = df['stay_ID']
        
        # Convert datetime columns
        print("Converting datetime fields...")
        df['book_in_time'] = pd.to_datetime(df['book_in_date_time_first'], errors='coerce')
        df['book_out_time'] = pd.to_datetime(df['book_out_date_time_last'], errors='coerce')
        
        print(f"\nLoaded {len(df):,} records")
        print(f"Children (age < 18): {len(df[df['age'] < 18]):,}")
        print(f"Adults (age >= 18): {len(df[df['age'] >= 18]):,}")
        
        return df
    
    def match_families_vectorized(self, df):
        """
        Optimized vectorized matching using pandas merge operations.
        Much faster for large datasets.
        
        Args:
            df: DataFrame containing detention records
            
        Returns:
            DataFrame with all potential matches
        """
        print("\n" + "="*60)
        print("Starting Family Matching (Vectorized Algorithm)")
        print("="*60)
        
        # Step 1: Separate children and adults
        print("\n[1/6] Separating children and adults...")
        children = df[df['age'] < 18].copy()
        adults = df[df['age'] >= 18].copy()
        
        print(f"  → {len(children):,} children")
        print(f"  → {len(adults):,} adults")
        
        if len(children) == 0:
            print("\n⚠️  No children found in dataset!")
            return pd.DataFrame()
        
        if len(adults) == 0:
            print("\n⚠️  No adults found in dataset!")
            return pd.DataFrame()
        
        # Step 2: Add prefixes to distinguish child vs parent columns
        print("\n[2/6] Preparing data for merge...")
        children_cols = children[['stay_id', 'age', 'gender', 'citizenship', 'facility', 
                                   'book_in_time', 'book_out_time']].copy()
        children_cols.columns = ['child_' + col for col in children_cols.columns]
        
        adults_cols = adults[['stay_id', 'age', 'gender', 'citizenship', 'facility',
                               'book_in_time', 'book_out_time']].copy()
        adults_cols.columns = ['parent_' + col for col in adults_cols.columns]
        
        # Step 3: Merge on facility and citizenship (creates all potential pairs)
        print("\n[3/6] Merging children with adults (same facility & citizenship)...")
        print("  This creates all potential child-parent pairs...")
        
        matches = children_cols.merge(
            adults_cols,
            left_on=['child_facility', 'child_citizenship'],
            right_on=['parent_facility', 'parent_citizenship'],
            how='inner'
        )
        
        print(f"  → {len(matches):,} potential pairs after facility/citizenship filter")
        
        if len(matches) == 0:
            print("\n⚠️  No matches found after facility/citizenship filtering")
            return pd.DataFrame()
        
        # Step 4: Calculate time differences
        print("\n[4/6] Calculating time differences...")
        matches['time_diff_minutes'] = (
            (matches['parent_book_in_time'] - matches['child_book_in_time'])
            .abs()
            .dt.total_seconds() / 60
        )
        
        matches['bookout_diff_minutes'] = (
            (matches['parent_book_out_time'] - matches['child_book_out_time'])
            .abs()
            .dt.total_seconds() / 60
        )
        
        # Step 5: Apply filters
        print("\n[5/6] Applying filters...")
        
        # Filter by time window
        time_window_minutes = self.time_window_hours * 60
        matches = matches[matches['time_diff_minutes'] <= time_window_minutes]
        print(f"  → {len(matches):,} pairs within {self.time_window_hours}h time window")
        
        if len(matches) == 0:
            print("\n⚠️  No matches found within time window")
            return pd.DataFrame()
        
        # Filter by age gap
        matches['age_gap'] = matches['parent_age'] - matches['child_age']
        matches = matches[
            (matches['age_gap'] >= self.min_parent_age_gap) & 
            (matches['age_gap'] <= self.max_parent_age_gap)
        ]
        print(f"  → {len(matches):,} pairs with reasonable age gap ({self.min_parent_age_gap}-{self.max_parent_age_gap} years)")
        
        if len(matches) == 0:
            print("\n⚠️  No matches found with reasonable age gaps")
            return pd.DataFrame()
        
        # Step 6: Calculate confidence scores
        print("\n[6/6] Calculating confidence scores...")
        matches['confidence'] = self._calculate_confidence_vectorized(matches)
        
        # Sort by confidence
        matches = matches.sort_values('confidence', ascending=False)
        
        print(f"\n✓ Matching complete! Found {len(matches):,} potential parent-child pairs")
        
        return matches
    
    def _calculate_confidence_vectorized(self, matches):
        """Calculate confidence scores for all matches at once."""
        score = np.zeros(len(matches))
        
        # Time proximity scoring
        score += np.where(matches['time_diff_minutes'] < 30, 40,
                 np.where(matches['time_diff_minutes'] < 120, 30,
                 np.where(matches['time_diff_minutes'] < 240, 20, 10)))
        
        # Book-out synchronization
        bookout_valid = matches['bookout_diff_minutes'].notna()
        score += np.where(bookout_valid & (matches['bookout_diff_minutes'] < 30), 30,
                 np.where(bookout_valid & (matches['bookout_diff_minutes'] < 120), 20,
                 np.where(bookout_valid & (matches['bookout_diff_minutes'] < 360), 10, 0)))
        
        # Count potential parents per child
        parent_counts = matches.groupby('child_stay_id').size()
        matches['parent_count'] = matches['child_stay_id'].map(parent_counts)
        score += np.where(matches['parent_count'].isin([1, 2]), 30, 0)
        
        # Cap at 100
        score = np.minimum(score, 100)
        
        return score
    
    def summarize_by_child(self, matches_df):
        """
        Convert the flat matches DataFrame into a summary grouped by child.
        
        Returns:
            List of dictionaries, one per child with their potential parents
        """
        if len(matches_df) == 0:
            return []
        
        print("\nGrouping matches by child...")
        
        child_groups = matches_df.groupby('child_stay_id')
        summaries = []
        
        for child_id, group in child_groups:
            # Get child info (same for all rows in group)
            first_row = group.iloc[0]
            
            # Calculate overall confidence (max confidence from any parent match)
            overall_confidence = int(group['confidence'].max())
            
            summary = {
                'child_id': child_id,
                'child_age': int(first_row['child_age']),
                'child_gender': first_row['child_gender'],
                'child_citizenship': first_row['child_citizenship'],
                'child_facility': first_row['child_facility'],
                'child_book_in': first_row['child_book_in_time'],
                'child_book_out': first_row['child_book_out_time'],
                'num_potential_parents': len(group),
                'confidence': overall_confidence,
                'potential_parents': []
            }
            
            # Add each potential parent
            for _, parent_row in group.iterrows():
                summary['potential_parents'].append({
                    'parent_id': parent_row['parent_stay_id'],
                    'parent_age': int(parent_row['parent_age']),
                    'parent_gender': parent_row['parent_gender'],
                    'parent_book_in': parent_row['parent_book_in_time'],
                    'parent_book_out': parent_row['parent_book_out_time'],
                    'time_diff_minutes': float(parent_row['time_diff_minutes']),
                    'bookout_diff_minutes': float(parent_row['bookout_diff_minutes']) if pd.notna(parent_row['bookout_diff_minutes']) else None,
                    'confidence': int(parent_row['confidence'])
                })
            
            summaries.append(summary)
        
        # Sort by confidence
        summaries.sort(key=lambda x: x['confidence'], reverse=True)
        
        print(f"✓ Created summaries for {len(summaries):,} children")
        
        return summaries
    
    def generate_report(self, matches_df):
        """Generate a summary report of matches from the vectorized output."""
        
        if len(matches_df) == 0:
            print("\n⚠️  No matches to report")
            return
        
        # Get unique children
        unique_children = matches_df['child_stay_id'].nunique()
        
        # Confidence breakdown
        high_conf = len(matches_df[matches_df['confidence'] >= 70])
        med_conf = len(matches_df[(matches_df['confidence'] >= 40) & (matches_df['confidence'] < 70)])
        low_conf = len(matches_df[matches_df['confidence'] < 40])
        
        print("\n" + "="*60)
        print("FAMILY MATCHING REPORT")
        print("="*60)
        print(f"\nTotal children with potential matches: {unique_children:,}")
        print(f"Total potential parent-child pairs: {len(matches_df):,}")
        print(f"\nConfidence Distribution (by pair):")
        print(f"  High confidence (70-100%): {high_conf:,} pairs")
        print(f"  Medium confidence (40-69%): {med_conf:,} pairs")
        print(f"  Low confidence (<40%): {low_conf:,} pairs")
        
        # Top facilities
        print(f"\nTop 5 Facilities with Matches:")
        top_facilities = matches_df['child_facility'].value_counts().head()
        for facility, count in top_facilities.items():
            print(f"  {facility}: {count:,} pairs")
        
        # Top citizenship countries
        print(f"\nTop 5 Citizenship Countries:")
        top_countries = matches_df['child_citizenship'].value_counts().head()
        for country, count in top_countries.items():
            print(f"  {country}: {count:,} pairs")
        
        # Show top 10 matches
        print(f"\n" + "-"*60)
        print("TOP 10 HIGHEST CONFIDENCE MATCHES")
        print("-"*60)
        
        top_matches = matches_df.head(10)
        for i, (_, match) in enumerate(top_matches.iterrows(), 1):
            print(f"\n{i}. Child ID: {match['child_stay_id']}")
            print(f"   Age: {match['child_age']}, Gender: {match['child_gender']}")
            print(f"   Facility: {match['child_facility']}")
            print(f"   Citizenship: {match['child_citizenship']}")
            print(f"   Confidence: {match['confidence']:.0f}%")
            print(f"   → Matched with Parent ID: {match['parent_stay_id']}")
            print(f"      Age: {match['parent_age']}, Gender: {match['parent_gender']}")
            print(f"      Time difference: {match['time_diff_minutes']:.0f} minutes")
            if pd.notna(match['bookout_diff_minutes']):
                print(f"      Book-out difference: {match['bookout_diff_minutes']:.0f} minutes")


# Example usage:
if __name__ == "__main__":
    # Initialize matcher
    matcher = FamilyMatcher(
        time_window_hours=4,
        min_parent_age_gap=16,
        max_parent_age_gap=50
    )
    
    # Load data
    df = matcher.load_data('data/detentions2025.csv')
    
    # Find matches (vectorized - much faster!)
    matches_df = matcher.match_families_vectorized(df)
    
    # Generate report
    matcher.generate_report(matches_df)
    
    # Optional: Convert to child-centric format
    # child_summaries = matcher.summarize_by_child(matches_df)
    
    # Export matches
    if len(matches_df) > 0:
        print("\n" + "="*60)
        print("Exporting results...")
        
        # Save the flat DataFrame
        matches_df.to_csv('family_matches_detailed.csv', index=False)
        print(f"✓ Detailed matches saved to: family_matches_detailed.csv")
        
        # Save child-centric summary
        child_summaries = matcher.summarize_by_child(matches_df)
        if child_summaries:
            import json
            with open('family_matches_summary.json', 'w') as f:
                json.dump(child_summaries, f, indent=2, default=str)
            print(f"✓ Summary by child saved to: family_matches_summary.json")
        
        print("\n✓ All exports complete!")
    else:
        print("\nNo matches to export.")
```

```python
import pandas as pd
import numpy as np
from datetime import timedelta

class FamilySeparationAnalyzer:
    def __init__(self, matches_csv_path, original_data_path):
        """
        Initialize analyzer with the family matches and original detention data.
        
        Args:
            matches_csv_path: Path to the detailed matches CSV file
            original_data_path: Path to original detention data with detention_facility_codes_all
        """
        print("Loading data...")
        self.matches = pd.read_csv(matches_csv_path)
        self.original_data = pd.read_csv(original_data_path)
        self._prepare_data()
    
    def _prepare_data(self):
        """Prepare and merge the datasets."""
        print("Preparing match data...")
        
        # Convert datetime columns in matches
        self.matches['child_book_in_time'] = pd.to_datetime(self.matches['child_book_in_time'])
        self.matches['child_book_out_time'] = pd.to_datetime(self.matches['child_book_out_time'])
        self.matches['parent_book_in_time'] = pd.to_datetime(self.matches['parent_book_in_time'])
        self.matches['parent_book_out_time'] = pd.to_datetime(self.matches['parent_book_out_time'])
        
        # Prepare original data
        print("Loading facility history from original data...")
        facility_data = self.original_data[['stay_ID', 'detention_facility_codes_all', 'n_stints']].copy()
        facility_data.columns = ['stay_id', 'facilities_all', 'n_stints']
        
        # Merge facility info for children
        self.matches = self.matches.merge(
            facility_data,
            left_on='child_stay_id',
            right_on='stay_id',
            how='left',
            suffixes=('', '_child_merge')
        )
        self.matches.rename(columns={
            'facilities_all': 'child_facilities_all',
            'n_stints': 'child_n_stints'
        }, inplace=True)
        
        # Merge facility info for parents
        self.matches = self.matches.merge(
            facility_data,
            left_on='parent_stay_id',
            right_on='stay_id',
            how='left',
            suffixes=('', '_parent_merge')
        )
        self.matches.rename(columns={
            'facilities_all': 'parent_facilities_all',
            'n_stints': 'parent_n_stints'
        }, inplace=True)
        
        # Clean up merge columns
        cols_to_drop = [col for col in self.matches.columns if '_merge' in col or col == 'stay_id']
        self.matches.drop(columns=cols_to_drop, inplace=True, errors='ignore')
        
        print(f"✓ Loaded {len(self.matches):,} potential parent-child pairs")
    
    def analyze_separation(self):
        """
        Determine if families stayed together or were separated based on:
        1. Facility movement history (detention_facility_codes_all)
        2. Release timing
        3. Number of facility transfers
        
        Returns:
            DataFrame with separation analysis
        """
        print("\nAnalyzing family separation patterns...")
        
        df = self.matches.copy()
        
        # === TIMING ANALYSIS ===
        
        # Same book-in and book-out times
        df['same_bookin'] = df['time_diff_minutes'] == 0
        df['same_bookout'] = df['bookout_diff_minutes'] == 0
        df['exact_same_timing'] = df['same_bookin'] & df['same_bookout']
        
        # Book-out within 30 minutes (likely together)
        df['bookout_within_30min'] = df['bookout_diff_minutes'] <= 30
        
        # Calculate detention duration
        df['child_detention_days'] = (
            (df['child_book_out_time'] - df['child_book_in_time']).dt.total_seconds() / 86400
        )
        df['parent_detention_days'] = (
            (df['parent_book_out_time'] - df['parent_book_in_time']).dt.total_seconds() / 86400
        )
        
        # Check who left first
        df['child_left_first'] = df['child_book_out_time'] < df['parent_book_out_time']
        df['parent_left_first'] = df['parent_book_out_time'] < df['child_book_out_time']
        df['separation_days'] = abs(
            (df['child_book_out_time'] - df['parent_book_out_time']).dt.total_seconds() / 86400
        )
        
        # === FACILITY MOVEMENT ANALYSIS ===
        
        # Compare facility paths
        df['same_facility_path'] = df['child_facilities_all'] == df['parent_facilities_all']
        
        # Check if they shared ANY facilities
        def check_shared_facilities(row):
            if pd.isna(row['child_facilities_all']) or pd.isna(row['parent_facilities_all']):
                return None
            child_facilities = str(row['child_facilities_all']).split(',')
            parent_facilities = str(row['parent_facilities_all']).split(',')
            shared = set(child_facilities) & set(parent_facilities)
            return len(shared) > 0
        
        df['shared_any_facility'] = df.apply(check_shared_facilities, axis=1)
        
        # Count number of shared facilities
        def count_shared_facilities(row):
            if pd.isna(row['child_facilities_all']) or pd.isna(row['parent_facilities_all']):
                return 0
            child_facilities = str(row['child_facilities_all']).split(',')
            parent_facilities = str(row['parent_facilities_all']).split(',')
            return len(set(child_facilities) & set(parent_facilities))
        
        df['num_shared_facilities'] = df.apply(count_shared_facilities, axis=1)
        
        # Multiple transfers indicator
        df['child_transferred'] = df['child_n_stints'] > 1
        df['parent_transferred'] = df['parent_n_stints'] > 1
        df['either_transferred'] = df['child_transferred'] | df['parent_transferred']
        
        # === SEPARATION STATUS CLASSIFICATION ===
        
        df['separation_status'] = 'Unknown'
        df['separation_confidence'] = 'Low'
        
        # DEFINITELY TOGETHER: Same timing + same facility path
        mask_together = (
            df['exact_same_timing'] & 
            df['same_facility_path'] & 
            (df['num_shared_facilities'] > 0)
        )
        df.loc[mask_together, 'separation_status'] = 'Stayed Together'
        df.loc[mask_together, 'separation_confidence'] = 'High'
        
        # LIKELY TOGETHER: Close timing + shared all facilities
        mask_likely_together = (
            ~mask_together &
            df['same_bookin'] &
            df['bookout_within_30min'] &
            df['same_facility_path']
        )
        df.loc[mask_likely_together, 'separation_status'] = 'Likely Together'
        df.loc[mask_likely_together, 'separation_confidence'] = 'Medium'
        
        # SEPARATED - DIFFERENT FACILITIES: No shared facilities
        mask_diff_facilities = (
            ~mask_together &
            ~mask_likely_together &
            (df['num_shared_facilities'] == 0) &
            df['shared_any_facility'].notna()
        )
        df.loc[mask_diff_facilities, 'separation_status'] = 'Separated - Different Facilities'
        df.loc[mask_diff_facilities, 'separation_confidence'] = 'High'
        
        # SEPARATED - TRANSFERRED APART: Started together, then different paths
        mask_transferred_apart = (
            ~mask_together &
            ~mask_likely_together &
            ~mask_diff_facilities &
            (df['num_shared_facilities'] > 0) &
            ~df['same_facility_path'] &
            df['either_transferred']
        )
        df.loc[mask_transferred_apart, 'separation_status'] = 'Separated - Transferred Apart'
        df.loc[mask_transferred_apart, 'separation_confidence'] = 'High'
        
        # SEPARATED - DIFFERENT RELEASE TIMES: Same facilities but released far apart
        mask_time_separation = (
            ~mask_together &
            ~mask_likely_together &
            ~mask_diff_facilities &
            ~mask_transferred_apart &
            df['same_facility_path'] &
            (df['separation_days'] > 7)  # Released more than a week apart
        )
        df.loc[mask_time_separation, 'separation_status'] = 'Separated - Released Apart'
        df.loc[mask_time_separation, 'separation_confidence'] = 'Medium'
        
        # STILL DETAINED
        mask_still_detained = (
            df['child_book_out_time'].isna() | 
            df['parent_book_out_time'].isna()
        )
        df.loc[mask_still_detained, 'separation_status'] = 'Still Detained'
        df.loc[mask_still_detained, 'separation_confidence'] = 'N/A'
        
        # UNCLEAR: Everything else
        mask_unclear = df['separation_status'] == 'Unknown'
        df.loc[mask_unclear, 'separation_status'] = 'Unclear - Need More Data'
        df.loc[mask_unclear, 'separation_confidence'] = 'Low'
        
        print(f"✓ Analysis complete")
        
        return df
    
    def generate_separation_report(self, analyzed_df):
        """Generate comprehensive separation statistics."""
        
        print("\n" + "="*70)
        print("FAMILY SEPARATION ANALYSIS REPORT")
        print("="*70)
        
        total_pairs = len(analyzed_df)
        unique_children = analyzed_df['child_stay_id'].nunique()
        
        print(f"\nTotal potential parent-child pairs analyzed: {total_pairs:,}")
        print(f"Unique children: {unique_children:,}")
        
        # Separation status breakdown
        print("\n" + "-"*70)
        print("SEPARATION STATUS BREAKDOWN")
        print("-"*70)
        
        status_counts = analyzed_df['separation_status'].value_counts()
        status_pcts = analyzed_df['separation_status'].value_counts(normalize=True) * 100
        
        for status in status_counts.index:
            count = status_counts[status]
            pct = status_pcts[status]
            
            # Get confidence for this status
            conf_counts = analyzed_df[analyzed_df['separation_status'] == status]['separation_confidence'].value_counts()
            conf_str = ", ".join([f"{c}: {v}" for c, v in conf_counts.items()])
            
            print(f"{status:35s}: {count:6,} ({pct:5.1f}%) [{conf_str}]")
        
        # Facility overlap analysis
        print("\n" + "-"*70)
        print("FACILITY SHARING ANALYSIS")
        print("-"*70)
        
        same_path = analyzed_df['same_facility_path'].sum()
        same_path_pct = (same_path / len(analyzed_df)) * 100
        print(f"Same exact facility path: {same_path:,} pairs ({same_path_pct:.1f}%)")
        
        shared_any = analyzed_df['shared_any_facility'].sum()
        shared_any_pct = (shared_any / analyzed_df['shared_any_facility'].notna().sum()) * 100
        print(f"Shared at least one facility: {shared_any:,} pairs ({shared_any_pct:.1f}%)")
        
        no_shared = (~analyzed_df['shared_any_facility']).sum()
        no_shared_pct = (no_shared / analyzed_df['shared_any_facility'].notna().sum()) * 100
        print(f"No shared facilities: {no_shared:,} pairs ({no_shared_pct:.1f}%)")
        
        print(f"\nAverage shared facilities: {analyzed_df['num_shared_facilities'].mean():.2f}")
        print(f"Max shared facilities: {analyzed_df['num_shared_facilities'].max():.0f}")
        
        # Transfer analysis
        print("\n" + "-"*70)
        print("TRANSFER ANALYSIS")
        print("-"*70)
        
        child_transfers = analyzed_df['child_transferred'].sum()
        parent_transfers = analyzed_df['parent_transferred'].sum()
        either_transfers = analyzed_df['either_transferred'].sum()
        
        print(f"Children transferred between facilities: {child_transfers:,} ({child_transfers/len(analyzed_df)*100:.1f}%)")
        print(f"Parents transferred between facilities: {parent_transfers:,} ({parent_transfers/len(analyzed_df)*100:.1f}%)")
        print(f"At least one transferred: {either_transfers:,} ({either_transfers/len(analyzed_df)*100:.1f}%)")
        
        # Detailed statistics for each category
        print("\n" + "-"*70)
        print("DETAILED STATISTICS BY CATEGORY")
        print("-"*70)
        
        categories = [
            'Stayed Together',
            'Likely Together',
            'Separated - Different Facilities',
            'Separated - Transferred Apart',
            'Separated - Released Apart'
        ]
        
        for category in categories:
            subset = analyzed_df[analyzed_df['separation_status'] == category]
            if len(subset) == 0:
                continue
                
            print(f"\n{category} ({len(subset):,} pairs):")
            print(f"  Avg child detention: {subset['child_detention_days'].mean():.1f} days")
            print(f"  Avg parent detention: {subset['parent_detention_days'].mean():.1f} days")
            
            if category.startswith('Separated'):
                print(f"  Avg separation time: {subset['separation_days'].mean():.1f} days")
                print(f"  Max separation time: {subset['separation_days'].max():.1f} days")
            
            print(f"  Shared facilities avg: {subset['num_shared_facilities'].mean():.1f}")
        
        # Top facilities with separations
        print("\n" + "-"*70)
        print("TOP 10 FACILITIES WITH MOST SEPARATION CASES")
        print("-"*70)
        
        separated = analyzed_df[analyzed_df['separation_status'].str.contains('Separated')]
        if len(separated) > 0:
            top_facilities = separated['child_facility'].value_counts().head(10)
            for i, (facility, count) in enumerate(top_facilities.items(), 1):
                print(f"{i:2d}. {facility:15s}: {count:4,} separation cases")
    
    def export_results(self, analyzed_df, output_path='family_separation_analysis.csv'):
        """Export the full analysis to CSV."""
        print(f"\nExporting results to {output_path}...")
        
        # Select key columns for export
        export_cols = [
            'child_stay_id', 'child_age', 'child_gender', 'child_citizenship', 'child_facility',
            'child_book_in_time', 'child_book_out_time', 'child_detention_days',
            'child_facilities_all', 'child_n_stints',
            'parent_stay_id', 'parent_age', 'parent_gender', 'parent_citizenship', 'parent_facility',
            'parent_book_in_time', 'parent_book_out_time', 'parent_detention_days',
            'parent_facilities_all', 'parent_n_stints',
            'time_diff_minutes', 'bookout_diff_minutes', 'separation_days',
            'same_facility_path', 'num_shared_facilities', 'shared_any_facility',
            'child_transferred', 'parent_transferred',
            'separation_status', 'separation_confidence',
            'confidence'
        ]
        
        # Only export columns that exist
        export_cols = [col for col in export_cols if col in analyzed_df.columns]
        
        analyzed_df[export_cols].to_csv(output_path, index=False)
        print(f"✓ Exported {len(analyzed_df):,} records")
        
        # Also create a summary by child
        print("\nCreating summary by child...")
        summary = self._create_child_summary(analyzed_df)
        summary_path = output_path.replace('.csv', '_by_child.csv')
        summary.to_csv(summary_path, index=False)
        print(f"✓ Exported child summary to {summary_path}")
    
    def _create_child_summary(self, analyzed_df):
        """Create one row per child with their separation status."""
        summary = analyzed_df.groupby('child_stay_id').agg({
            'child_age': 'first',
            'child_gender': 'first',
            'child_citizenship': 'first',
            'child_facility': 'first',
            'child_facilities_all': 'first',
            'child_detention_days': 'first',
            'child_n_stints': 'first',
            'parent_stay_id': 'count',  # Number of potential parents
            'separation_status': lambda x: x.value_counts().index[0],  # Most common status
            'num_shared_facilities': 'mean',
            'separation_days': 'mean'
        }).reset_index()
        
        summary.columns = [
            'child_stay_id', 'age', 'gender', 'citizenship', 'initial_facility',
            'all_facilities', 'detention_days', 'num_transfers', 'num_potential_parents',
            'separation_status', 'avg_shared_facilities', 'avg_separation_days'
        ]
        
        return summary


# Example usage:
if __name__ == "__main__":
    # Initialize analyzer with both datasets
    analyzer = FamilySeparationAnalyzer(
        matches_csv_path='family_matches_detailed.csv',
        original_data_path='data/detentions2025.csv'
    )
    
    # Analyze separation
    results = analyzer.analyze_separation()
    
    # Generate report
    analyzer.generate_separation_report(results)
    
    # Export results
    analyzer.export_results(results, 'family_separation_analysis.csv')
    
    print("\n✓ Analysis complete!")
```
